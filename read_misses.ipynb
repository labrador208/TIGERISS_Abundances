```python

import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('original_misses_log_deg090.txt', sep='\t', header=None, names=['x', 'y', 'layer'])

x_original=df['x']
y_original=df['y']


df2 = pd.read_csv('rotated_misses_log_deg090.txt', sep='\t', header=None, names=['x', 'y', 'layer'])
x_rotated=df2['x']
y_rotated=df2['y']
```

```python
plt.scatter(x_original, y_original, s=0.2)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Original Misses')
#plt.xlim(-200, 200)
#plt.ylim(-200,200)
plt.show()
```

```python
plt.scatter(x_rotated, y_rotated, s=0.2)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Rotated Misses')
#plt.xlim(-200, 200)
#plt.ylim(-1000,1000)
plt.show()
```
